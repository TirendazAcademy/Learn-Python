# Metot Türleri

```python
class DemoClass:
    # Instance metot 
    def instance_method(self):
        return ("instance method called", self)
    # Class metot
    @classmethod
    def class_method(cls):
        return("class method called", cls)
    # Statik metot
    @staticmethod
    def static_method():
        return("static method called")
```

```python
obj = DemoClass()
```

```python
print(obj.instance_method())
print(DemoClass.class_method())
print(DemoClass.static_method())
```

```python
class Kedi(Hayvan):
    def ses_cikar(self):
        print("Miyav")
```

```python
kedi_obj.sayi()
```

## Üst sınıfa yönlendirme (Delegating to Superclass )

```python
class Base:
    def greet(self, name):
        print("Merhaba", name)
```

```python
class Sub(Base):
    def greet(self, name):
        print("Hoş geldin", end=" ")
        Base.greet(self,name)
```

```python
s = Sub()
```

```python
s.greet("Ali")
```

```python
class Base:
    def greet(self, name):
        print("Merhaba", name)
```

```python
class Sub(Base):
    def greet(self, name):
        print("Hoş geldin", end=" ")
        super().greet(name)
```

```python
s = Sub()
```

```python
s.greet("Ali")
```

```python
class Base:
    def __init__(self):
        self.x = 10
```

```python
class Sub(Base):
    def __init__(self):
        super().__init__()
        self.y=20
```

```python
s = Sub()
```

```python
print(s.x, s.y)
```

# Uygulama

```python
class Arac:
    def __init__(self, marka):
        self.marka = marka
    def hareket_et(self):
        print(f"{self.marka} aracı hareket ediyor")
```

```python
class Araba(Arac):
    def hareket_et(self):
        print("🚗 Araba çalıtırılıyor...")
        super().hareket_et()
```

```python
class Bisiklet(Arac):
    def hareket_et(self):
        print("🚴 Bisiklet pedalları dönüyor...")
        super().hareket_et()
```

```python
class ElektrikliAraba(Araba, Arac):
    def hareket_et(self):
        print("🔋 Elektirikli araba sessizce çalışıyor...")
        super().hareket_et()
```

```python
a = Arac("Genel Araç")
a.hareket_et()
```

```python
b = Araba("Toyota")
b.hareket_et()
```

```python
bi = Bisiklet("Bianchi")
bi.hareket_et()
```

```python
e = ElektrikliAraba("Tesla")
e.hareket_et()
```

```python
print("\nElektrikliAraba MRO:")
print(ElektrikliAraba.__mro__)
```
